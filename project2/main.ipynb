```python
def plotting(var, num):
    # plt.figure(figsize=(10,8))
    plt.subplot(2,2,num)
    sns.histplot(df[var], kde=True)
    plt.tight_layout()
```

```python
plotting("Age", 1)
plotting("RestingBP", 2)
plotting("Cholesterol", 3)
plotting("MaxHR", 4)
```
